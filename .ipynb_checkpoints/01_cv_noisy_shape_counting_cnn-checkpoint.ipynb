```python
import io, random, time, warnings
from pathlib import Path
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from PIL import Image, ImageFilter
from sklearn.model_selection import GroupShuffleSplit
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
import torch
import torch.nn as nn
import torch.nn.functional as F
from torch.utils.data import Dataset, DataLoader
warnings.filterwarnings("ignore")
SEED = 42
random.seed(SEED)
np.random.seed(SEED)
torch.manual_seed(SEED)
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
torch.set_num_threads(max(1, min(8, (torch.get_num_threads()))))
print("device:", device)
print("torch:", torch.__version__)
```

```python
data_dir = Path("data")
data_dir.mkdir(exist_ok=True)
cv_file = data_dir / "shape_counting.parquet"
url = "https://huggingface.co/api/datasets/nooranis/shape-counting-dataset/parquet/default/train/0.parquet"
if not cv_file.exists():
    import requests
    response = requests.get(url, timeout=180)
    response.raise_for_status()
    cv_file.write_bytes(response.content)
df = pd.read_parquet(cv_file).reset_index(drop=True)
print("shape:", df.shape)
print("base images:", df.image_id.nunique())
print("noise types:", df.noise_type.nunique())
df[["noise_type", "count_squares", "count_triangles", "count_stars", "total_shapes", "difficulty"]].head(10)
```

```python
fig, axes = plt.subplots(1, 3, figsize=(15, 4))
df.total_shapes.hist(bins=25, ax=axes[0], color="steelblue")
axes[0].set_title("Total-count distribution")
df.noise_type.value_counts().plot.bar(ax=axes[1], color="darkorange")
axes[1].set_title("Noise variants")
df.difficulty.value_counts().plot.bar(ax=axes[2], color="seagreen")
axes[2].set_title("Difficulty")
plt.tight_layout()
plt.show()
print(df.groupby("difficulty").total_shapes.agg(["count", "min", "mean", "max"]).round(2))
```

```python
example_id = int(df[df.bucket == 3].image_id.iloc[0])
example = df[df.image_id == example_id]
fig, axes = plt.subplots(2, 4, figsize=(14, 7))
for ax, (_, row) in zip(axes.ravel(), example.iterrows()):
    image = Image.open(io.BytesIO(row.image["bytes"])).convert("L")
    ax.imshow(image, cmap="gray", vmin=0, vmax=255)
    ax.set_title(row.noise_type)
    ax.axis("off")
plt.suptitle(f"image_id={example_id}, total={int(example.total_shapes.iloc[0])}")
plt.tight_layout()
plt.show()
```

```python
groups = df.image_id.to_numpy()
gss1 = GroupShuffleSplit(n_splits=1, train_size=0.70, random_state=SEED)
train_idx, temp_idx = next(gss1.split(df, groups=groups))
gss2 = GroupShuffleSplit(n_splits=1, train_size=0.50, random_state=SEED + 1)
val_rel, test_rel = next(gss2.split(df.iloc[temp_idx], groups=groups[temp_idx]))
val_idx = temp_idx[val_rel]
test_idx = temp_idx[test_rel]
train_groups = set(groups[train_idx])
val_groups = set(groups[val_idx])
test_groups = set(groups[test_idx])
print("rows:", len(train_idx), len(val_idx), len(test_idx))
print("base images:", len(train_groups), len(val_groups), len(test_groups))
print("overlap:", len(train_groups & val_groups), len(train_groups & test_groups), len(val_groups & test_groups))
print("target means:", *(round(df.iloc[i].total_shapes.mean(), 3) for i in [train_idx, val_idx, test_idx]))
```

```python
size = 64
gray = torch.empty((len(df), 1, size, size), dtype=torch.uint8)
t0 = time.time()
for i, item in enumerate(df.image):
    image = Image.open(io.BytesIO(item["bytes"])).convert("L").resize((size, size), Image.Resampling.BILINEAR)
    image = image.filter(ImageFilter.MedianFilter(3))
    gray[i, 0] = torch.from_numpy(np.asarray(image, dtype=np.uint8).copy())
sobel_x = torch.tensor([[-1., 0., 1.], [-2., 0., 2.], [-1., 0., 1.]]).view(1, 1, 3, 3)
sobel_y = sobel_x.transpose(2, 3)
edge = torch.empty_like(gray)
for start in range(0, len(df), 512):
    x = gray[start:start + 512].float() / 255
    gx = F.conv2d(x, sobel_x, padding=1)
    gy = F.conv2d(x, sobel_y, padding=1)
    magnitude = torch.sqrt(gx.square() + gy.square()).clamp(0, 2) / 2
    edge[start:start + 512] = (magnitude * 255).to(torch.uint8)
images = torch.cat([255 - gray, edge], dim=1)
targets = torch.tensor(df.total_shapes.to_numpy(), dtype=torch.float32)
print("tensor:", tuple(images.shape), images.dtype, "seconds:", round(time.time() - t0, 2))
```

```python
row_ids = [int(train_idx[0]), int(val_idx[0]), int(test_idx[0])]
fig, axes = plt.subplots(3, 3, figsize=(9, 9))
for r, idx in enumerate(row_ids):
    raw = Image.open(io.BytesIO(df.iloc[idx].image["bytes"])).convert("L")
    axes[r, 0].imshow(raw, cmap="gray")
    axes[r, 1].imshow(images[idx, 0], cmap="gray")
    axes[r, 2].imshow(images[idx, 1], cmap="magma")
    axes[r, 0].set_ylabel(f"count={int(targets[idx])}")
for ax, title in zip(axes[0], ["Raw", "Median foreground", "Sobel magnitude"]):
    ax.set_title(title)
for ax in axes.ravel():
    ax.set_xticks([])
    ax.set_yticks([])
plt.tight_layout()
plt.show()
```

```python
class ShapeDataset(Dataset):
    def __init__(self, indices, augment=False):
        self.indices = torch.as_tensor(indices, dtype=torch.long)
        self.augment = augment
    def __len__(self):
        return len(self.indices)
    def __getitem__(self, index):
        j = self.indices[index]
        x = images[j].float() / 255
        if self.augment:
            if torch.rand(()) < 0.5:
                x = x.flip(-1)
            if torch.rand(()) < 0.5:
                x = x.flip(-2)
            x = torch.rot90(x, int(torch.randint(0, 4, ()).item()), (-2, -1))
            if torch.rand(()) < 0.35:
                x = (x + torch.randn_like(x) * 0.015).clamp(0, 1)
        return x, targets[j]
train_loader = DataLoader(ShapeDataset(train_idx, True), batch_size=256, shuffle=True, num_workers=0)
val_loader = DataLoader(ShapeDataset(val_idx), batch_size=512, shuffle=False, num_workers=0)
test_loader = DataLoader(ShapeDataset(test_idx), batch_size=512, shuffle=False, num_workers=0)
xb, yb = next(iter(train_loader))
print("batch:", tuple(xb.shape), tuple(yb.shape), yb.min().item(), yb.max().item())
```

```python
class CountCNN(nn.Module):
    def __init__(self):
        super().__init__()
        self.features = nn.Sequential(
            nn.Conv2d(2, 24, 5, padding=2), nn.BatchNorm2d(24), nn.SiLU(),
            nn.Conv2d(24, 32, 3, stride=2, padding=1), nn.BatchNorm2d(32), nn.SiLU(),
            nn.Conv2d(32, 48, 3, stride=2, padding=1), nn.BatchNorm2d(48), nn.SiLU(),
            nn.Conv2d(48, 64, 3, stride=2, padding=1), nn.BatchNorm2d(64), nn.SiLU(),
            nn.Conv2d(64, 96, 3, stride=2, padding=1), nn.BatchNorm2d(96), nn.SiLU(),
            nn.AdaptiveAvgPool2d(1)
        )
        self.head = nn.Sequential(nn.Linear(98, 64), nn.SiLU(), nn.Dropout(0.10), nn.Linear(64, 1))
    def forward(self, x):
        z = self.features(x).flatten(1)
        stats = x.mean(dim=(2, 3))
        return self.head(torch.cat([z, stats], dim=1)).squeeze(1)
model = CountCNN().to(device)
print(model)
print("parameters:", sum(p.numel() for p in model.parameters()))
print("forward:", tuple(model(xb[:4].to(device)).shape))
```

```python
target_mean = targets[train_idx].mean().item()
target_std = targets[train_idx].std().item()
optimizer = torch.optim.AdamW(model.parameters(), lr=2e-3, weight_decay=2e-4)
scheduler = torch.optim.lr_scheduler.ReduceLROnPlateau(optimizer, mode="min", factor=0.5, patience=2)
history = []
best = float("inf")
best_state = None
patience = 0
for epoch in range(1, 21):
    model.train()
    train_sum = 0.0
    for x, y in train_loader:
        x, y = x.to(device), y.to(device)
        pred_z = model(x)
        loss = F.smooth_l1_loss(pred_z, (y - target_mean) / target_std, beta=0.35)
        optimizer.zero_grad()
        loss.backward()
        nn.utils.clip_grad_norm_(model.parameters(), 3.0)
        optimizer.step()
        train_sum += loss.item() * len(y)
    model.eval()
    val_pred, val_true = [], []
    with torch.no_grad():
        for x, y in val_loader:
            pred = model(x.to(device)).cpu() * target_std + target_mean
            val_pred.append(pred)
            val_true.append(y)
    vp = torch.cat(val_pred).numpy()
    vt = torch.cat(val_true).numpy()
    val_mae = mean_absolute_error(vt, vp)
    train_loss = train_sum / len(train_loader.dataset)
    history.append((epoch, train_loss, val_mae))
    scheduler.step(val_mae)
    if val_mae < best - 0.01:
        best = val_mae
        best_state = {k: v.detach().cpu().clone() for k, v in model.state_dict().items()}
        patience = 0
    else:
        patience += 1
    print(f"epoch={epoch:02d} train_loss={train_loss:.4f} val_mae={val_mae:.3f} lr={optimizer.param_groups[0]['lr']:.1e}")
    if patience >= 5 and epoch >= 10:
        break
model.load_state_dict(best_state)
print("best validation MAE:", round(best, 4))
```

```python
hist = pd.DataFrame(history, columns=["epoch", "train_loss", "val_mae"])
fig, axes = plt.subplots(1, 2, figsize=(11, 4))
axes[0].plot(hist.epoch, hist.train_loss, marker="o")
axes[0].set_title("Training loss")
axes[1].plot(hist.epoch, hist.val_mae, marker="o", color="darkorange")
axes[1].set_title("Validation MAE")
for ax in axes:
    ax.set_xlabel("Epoch")
    ax.grid(alpha=.25)
plt.tight_layout()
plt.show()
```

```python
model.eval()
predictions, truths = [], []
with torch.no_grad():
    for x, y in test_loader:
        pred = model(x.to(device)).cpu() * target_std + target_mean
        predictions.append(pred)
        truths.append(y)
pred = torch.cat(predictions).numpy()
true = torch.cat(truths).numpy()
mae = mean_absolute_error(true, pred)
rmse = mean_squared_error(true, pred) ** 0.5
r2 = r2_score(true, pred)
within1 = np.mean(np.abs(pred - true) <= 1)
within2 = np.mean(np.abs(pred - true) <= 2)
rounded = np.mean(np.rint(pred) == true)
print(f"test MAE={mae:.3f}")
print(f"test RMSE={rmse:.3f}")
print(f"test R2={r2:.4f}")
print(f"within ±1={within1:.2%}")
print(f"within ±2={within2:.2%}")
print(f"rounded exact={rounded:.2%}")
```

```python
result = df.iloc[test_idx][["image_id", "noise_type", "difficulty", "total_shapes"]].copy()
result["prediction"] = pred
result["absolute_error"] = np.abs(result.prediction - result.total_shapes)
noise_report = result.groupby("noise_type").agg(n=("absolute_error", "size"), mae=("absolute_error", "mean"), within_2=("absolute_error", lambda x: np.mean(x <= 2))).sort_values("mae")
display(noise_report.round(3))
display(result.groupby("difficulty").absolute_error.agg(["count", "mean", "median", "max"]).round(3))
```

```python
fig, axes = plt.subplots(1, 2, figsize=(12, 5))
axes[0].scatter(true, pred, s=10, alpha=.35)
limits = [true.min(), true.max()]
axes[0].plot(limits, limits, "r--")
axes[0].set_xlabel("True count")
axes[0].set_ylabel("Predicted count")
axes[0].set_title("Predicted vs true")
worst = result.nlargest(6, "absolute_error")
for ax, (_, row) in zip([axes[1]], worst.head(1).iterrows()):
    idx = int(row.name)
    ax.imshow(Image.open(io.BytesIO(df.iloc[idx].image["bytes"])).convert("L"), cmap="gray")
    ax.set_title(f"Worst: true={row.total_shapes}, pred={row.prediction:.1f}, {row.noise_type}")
    ax.axis("off")
plt.tight_layout()
plt.show()
result.sort_values("absolute_error").head(10)
```
